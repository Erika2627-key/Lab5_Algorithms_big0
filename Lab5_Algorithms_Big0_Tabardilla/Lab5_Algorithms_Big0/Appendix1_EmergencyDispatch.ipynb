```python
# STUDENT-SPECIFIC INPUTS
LAST_NAME = "TABARDILLA" 
STUDENT_ID = "TUPM-26-7938"
FAVORITE_ARTIST = "ARTHUR NERY"

# Keep the complete ID as text, but use only its digits for numeric calculations.
STUDENT_ID_DIGITS = "".join(ch for ch in STUDENT_ID if ch.isdigit())

if not STUDENT_ID_DIGITS:
    raise ValueError("STUDENT_ID must contain at least one digit.")

SEED_DIGIT = int(STUDENT_ID_DIGITS[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID_DIGITS)
NAME_LENGTH = len(LAST_NAME.strip())
ARTIST_LENGTH = len(FAVORITE_ARTIST.strip())
ARTIST_SUM = sum(
    ord(ch.upper()) - ord("A") + 1
    for ch in FAVORITE_ARTIST
    if ch.isalpha() and "A" <= ch.upper() <= "Z"
)

print("STUDENT_ID:", STUDENT_ID)
print("SEED_DIGIT:", SEED_DIGIT)
print("ID_SUM:", ID_SUM)
print("NAME_LENGTH:", NAME_LENGTH)
print("ARTIST_LENGTH:", ARTIST_LENGTH)
print("ARTIST_SUM:", ARTIST_SUM)

```

```python
size = 10
original = sorted({
    ID_SUM + ARTIST_SUM + i * (NAME_LENGTH + 1)
    for i in range(size)
})
rotation = SEED_DIGIT % len(original)
rotated = original[rotation:] + original[:rotation]
target = ID_SUM + ARTIST_SUM

def linear_search(data, target):
    comparisons = 0
    for index, value in enumerate(data):
        comparisons += 1
        if value == target:
            return index, comparisons
    return -1, comparisons

original_index, original_comparisons = linear_search(original, target)
rotated_index, rotated_comparisons = linear_search(rotated, target)

print("Original Dataset:", original)
print("Rotated Dataset:", rotated)
print("Target Incident:", target)
print("Original Position (-1 means not found):", original_index)
print("Rotated Position (-1 means not found):", rotated_index)
print("Original Search Comparisons:", original_comparisons)
print("Rotated Search Comparisons:", rotated_comparisons)

print("\\nLinear Search Time Complexity: O(n)")
print("Linear Search Auxiliary Space: O(1)")
print("Rotation Time Complexity: O(n)")
```
