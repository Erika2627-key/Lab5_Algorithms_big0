```python
# STUDENT-SPECIFIC INPUTS
LAST_NAME = "TABARDILLA"
STUDENT_ID = "TUPM-26-7938"
FAVORITE_ARTIST = "ARTHUR NERY"

# Keep the complete ID as text, but use only its digits for numeric calculations.
STUDENT_ID_DIGITS = "".join(ch for ch in STUDENT_ID if ch.isdigit())

if not STUDENT_ID_DIGITS:
    raise ValueError("STUDENT_ID must contain at least one digit.")

SEED_DIGIT = int(STUDENT_ID_DIGITS[-1])
ID_SUM = sum(int(digit) for digit in STUDENT_ID_DIGITS)
NAME_LENGTH = len(LAST_NAME.strip())
ARTIST_LENGTH = len(FAVORITE_ARTIST.strip())
ARTIST_SUM = sum(
    ord(ch.upper()) - ord("A") + 1
    for ch in FAVORITE_ARTIST
    if ch.isalpha() and "A" <= ch.upper() <= "Z"
)

print("STUDENT_ID:", STUDENT_ID)
print("SEED_DIGIT:", SEED_DIGIT)
print("ID_SUM:", ID_SUM)
print("NAME_LENGTH:", NAME_LENGTH)
print("ARTIST_LENGTH:", ARTIST_LENGTH)
print("ARTIST_SUM:", ARTIST_SUM)
```

```python
dataset_a = sorted({
    SEED_DIGIT + i * ID_SUM
    for i in range(1, 6)
})
dataset_b = sorted({
    ARTIST_LENGTH + i * NAME_LENGTH
    for i in range(1, 6)
})

def merge_sorted(a, b):
    i = j = comparisons = 0
    merged = []
    while i < len(a) and j < len(b):
        comparisons += 1
        if a[i] <= b[j]:
            merged.append(a[i])
            i += 1
        else:
            merged.append(b[j])
            j += 1
    merged.extend(a[i:])
    merged.extend(b[j:])
    return merged, comparisons

consolidated, comparisons = merge_sorted(dataset_a, dataset_b)

print("Department A Dataset:", dataset_a)
print("Department B Dataset:", dataset_b)
print("Consolidated Dataset:", consolidated)
print("Records Before Consolidation:", len(dataset_a) + len(dataset_b))
print("Records After Consolidation:", len(consolidated))
print("Merge Comparisons:", comparisons)
print("Ordering Verified:", consolidated == sorted(consolidated))

print("\\nMerge Time Complexity: O(n + m)")
print("Auxiliary Space Complexity: O(n + m)")
```
